# **Matrix Multiplication **
## We will use tf.mantmul() method

1. Inner dimension must match --> (m,n)@(n,p). e.g:-
* (5,7)@(8,9) It will not work because innerdimension is different
* (5,7)@(7,9) It will work because inner dimension is the same




2. The result matrix has the shape of outer dimension. e.g:-
* A(m,n)B(n,p) = (m,p)
* (4,7)@(7,5) = (4,5)




```python
import tensorflow as tf
import numpy as np
```

```python
tensor_a = tf.constant([[1,2],[3,4],[4,5]],dtype=tf.int32)
tensor_b = tf.constant([[5,6],[7,8]],dtype=tf.int32)
print("Shape of A",tensor_a.shape)
print("Shape of B",tensor_b.shape)
```

```python
tensor_c = tf.matmul(tensor_a,tensor_b)
print(tensor_c)
print("Shape of C",tensor_c.shape)
```

```python
X = tf.constant([[1.,0.,-1.],
                 [2.,3.,1.]]
                 ,dtype=tf.float32)
X
```

```python
Y = tf.constant([[4.],
                 [5.],
                 [6.]
                 ],dtype=tf.float32)
Y
```

```python
print("X-Shape",X.shape)
print("Y-Shape",Y.shape)
```

```python
z=tf.matmul(X,Y)
z
```

```python
print(f"X-shape: {X.shape},Y-shape: {Y.shape}, z-shape: {z.shape}")
print(X.numpy())
print(Y.numpy())
print(z.numpy())
```

Transpose : Transpose swaps axes. For 2D matrix shape(m,n) become (n,m) after transpose

```python
tensor_one=tf.constant([
    [[1,2],[2,3]],
    [[4,5],[6,7]],], dtype=tf.int32)
tensor_one
print("Shape of tension_one: ",tensor_one.shape)
```

```python
tensor_one_transpose=tf.transpose(tensor_one)
print(tensor_one_transpose)
print("Shape of tension_one_transpose: ",tensor_one_transpose.shape)
```

```python
# 3D Tranpose with explicit axis permutation
tensor_two = tf.constant([
    [[1,2],[3,4]],
    [[5,6],[7,8]]], dtype=tf.int32)
print(tensor_two)
```

```python
# swap first and last axis (0,1,2)->(0,2,1)
t2_transpose = tf.transpose(tensor_two,perm=[0,2,1])
print(t2_transpose)
print("Shape of t2_transpose",t2_transpose.shape)
print("Shape of tensor_two",tensor_two.shape)
```

**Reshape**
# tf.reshape(tensor,new_shape)
# Change the shape without changing the data
**Rule** : The total number of element must stay the same


```python
tensor=tf.constant(
    [[1,2,3],
    [4,5,6]], dtype=tf.int32
)
print(tensor)
print("Shape of tensor",tensor.shape)
print("Number of element are: ",tf.size(tensor).numpy())
```

```python
tensor_r1=tf.reshape(tensor,shape=(3,2))
print(tensor_r1)
print("Shape of tensor_r1",tensor_r1.shape)
print("Number of element are: ",tf.size(tensor_r1).numpy())
```

```python
tensor_r2=tf.reshape(tensor,shape=(6,1))
print(tensor_r2)
print("Shape of tensor_r1",tensor_r2.shape)
print("Number of element are: ",tf.size(tensor_r2).numpy())
```

```python
tensor_r3=tf.reshape(tensor,shape=(1,6))
print(tensor_r3)
print("Shape of tensor_r1",tensor_r3.shape)
print("Number of element are: ",tf.size(tensor_r3).numpy())
```

```python
# infer dimension- if put -1 it will be decide the appropriate number to get total 6 element
tensor_r4=tf.reshape(tensor,shape=(-1,2))
print(tensor_r4)
print("Shape of tensor_r1",tensor_r4.shape)
print("Number of element are: ",tf.size(tensor_r4).numpy())
```

```python
tensor_r5=tf.reshape(tensor,shape=(-1,3))
print(tensor_r5)
print("Shape of tensor_r1",tensor_r5.shape)
print("Number of element are: ",tf.size(tensor_r5).numpy())
```
