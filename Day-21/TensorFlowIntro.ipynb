```python
# Find the shape,dimension,rank and size of tensors
print("Scalar Information")
print("Shape:", scalar.shape)
print("Dimension:", scalar.ndim)
print("Rank:", tf.rank(scalar).numpy())
print("Size:", tf.size(scalar))
```

```python
# Find the vector shape,dimension,rank and size
print("Vector Information")
print("Shape:", vector.shape)
print("Dimension:", vector.ndim)
print("Rank:", tf.rank(vector).numpy())
print("Size:", tf.size(vector))
```

```python
# Find the Matrix shape,dimension,rank and size
print("Matrix Information")
print("Shape:", matrix.shape)
print("Dimension:", matrix.ndim)
print("Rank:", tf.rank(matrix).numpy())
print("Size:", tf.size(matrix))
```

```python
# Find the Tensor shape,dimension,rank and size
print("Tensor Information")
print("Shape:", tensor.shape)
print("Dimension:", tensor.ndim)
print("Rank:", tf.rank(tensor).numpy())
print("Size:", tf.size(tensor))
```

```python
import numpy as np
numpy_Array=np.arange(1,25,dtype=np.int16)
numpy_Array
```

```python
tensor_numpy=tf.constant(numpy_Array,shape=(2,3,4))
tensor_numpy
```

```python
print("Tensor Information")
print("Shape:", tensor_numpy.shape)
print("Dimension:", tensor_numpy.ndim)
print("Rank:", tf.rank(tensor_numpy).numpy())
print("Size:", tf.size(tensor_numpy))
print("DataType of elements: ",tensor_numpy.dtype)
print("Element along axis 0 :",tensor_numpy.shape[0])
print("Element along axis 1 :",tensor_numpy.shape[1])
print("Element along axis 2 :",tensor_numpy.shape[2])
print("Total number of element: ", tf.size(tensor_numpy).numpy())
```

**Arithmetic Operation with Tensor**
*   Add
*   Sub
*   Multi
*   Div





```python
tensor_one=tf.constant([[11,9],[2,5]])
tensor_one
```

```python
# tensor_one+15 #one way
tf.add(tensor_one,15) #second way
```

```python
# tensor_one-5 #one way
tf.subtract(tensor_one,5) #second way
```

```python
# tensor_one*3 #one way
tf.multiply(tensor_one,3) #second way
```

```python
# tensor_one/2 # one way
tf.divide(tensor_one,2) #second way
```

```python
tensor_new=tf.square(tensor_one)
tensor_new
```
